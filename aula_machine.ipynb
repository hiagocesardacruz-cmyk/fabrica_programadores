```python
from sklearn.linear_model import Lasso

x = [[10_000], [15_000], [30_000], [42_000], [47_000] , [67_000]]
# Variável ALVO (Target)
y = [74_000, 80_000, 120_000, 150_000, 147_000, 210_000]

modelo = Lasso()
modelo.fit(x,y)

verba_marketing_previsao = [[1200]]
lucro_previsto = modelo.predict(verba_marketing_previsao)

print(f"Investindo R$ {verba_marketing_previsao[0][0]:,.2f} teremos lucro de R$ {lucro_previsto[0]:,.2f}")
```

```python
from sklearn.linear_model import Lasso

X_duracao = [[10], [15], [20], [25], [30], [35], [40], [50], [55], [60]]
y_preco = [10.00, 13.00, 14.00, 18.00, 19.00, 23.00, 24.00, 30.00, 32.00, 34.00]


modelo = Lasso()
modelo.fit(X_duracao, y_preco)


preco_45 = modelo.predict([[45]])[0]


preco_20 = modelo.predict([[20]])[0]
preco_60 = modelo.predict([[60]])[0]


print(f"Preço estimado para 45 minutos: R$ {preco_45:.2f}")
print(f"Preço estimado para 20 minutos: R$ {preco_20:.2f}")
print(f"Preço estimado para 60 minutos: R$ {preco_60:.2f}")
```
